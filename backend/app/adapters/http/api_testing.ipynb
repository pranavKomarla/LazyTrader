```python
# News API Testing/Sources

from newsapi import NewsApiClient

newsapi = NewsApiClient(api_key=NEWS_API_KEY)
sources = newsapi.get_sources() # /v2/top-headlines/sources
print(sources)

'''
Here are the English, stock/markets-focused sources the list of sources:

australian-financial-review — Australian Financial Review (EN, business)
bloomberg — Bloomberg (EN, business)
business-insider — Business Insider (EN, business)
financial-post — Financial Post (EN, business)
fortune — Fortune (EN, business)
the-wall-street-journal — The Wall Street Journal (EN, business)
reuters — Reuters (EN, general but very strong markets coverage)

'''
```

```python
# News API Testing

# Fetch top headlines -> /v2/top-headlines
top_headlines = newsapi.get_top_headlines(sources='bbc-news')
print(top_headlines)
```

```python
# News API Testing
# Fetch everything -> /v2/everything
all_articles = newsapi.get_everything(q='earnings', #earnings, ipo, merger, acquisition, etc.
                                      sources='bbc-news,bloomberg,business-insider,financial-post,fortune,the-wall-street-journal,reuters',
                                      domains='bbc.co.uk,techcrunch.com,bloomberg.com,businessinsider.com,financialpost.com,fortune.com,wsj.com,reuters.com,engadget.com,cnbc.com,theguardian.com,nytimes.com',
                                      from_param='2025-08-10',
                                      to='2025-08-11',
                                      language='en',
                                      sort_by='relevancy',
                                      page=1)
print(all_articles)


# It seems that the get_everything endpoint and the get_top_headlines endpoint are the same in that the model configuration is the same. 
# The to param has no limit, but the from param has a limit of 1 month back + 1 day.
```
